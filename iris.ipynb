```python
import pandas as pd
```

```python
df=pd.read_csv("C:/Users/Ibscai07.DGVC-MCA/Desktop/iris.csv")
df.head()
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['sepal_length']=df['sepal_length'].fillna(0)
```

```python
df.isnull().sum()
```

```python
mn=df['sepal_width'].mean()
mn
```

```python
df['sepal_width']=df['sepal_width'].fillna(mn)
```

```python
df.isnull().sum()
```

```python
md=df['petal_length'].median()
md
```

```python
df['petal_length']=df['petal_length'].fillna(md)
```

```python
df.isnull().sum()
```

```python
df['class']=df['class'].ffill()
df['class']=df['class'].bfill()
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
```

```python
x=df['sepal_length']
y=df['sepal_width']
x1=df['petal_length']
y1=df['petal_width']
plt.xlabel("sepal_length")
plt.ylabel("sepal_width")
plt.title("line plot Example1")
plt.plot(x,y,label="line-1",color='purple',linestyle='dashdot',linewidth=3,marker='D',markerfacecolor='red',markersize=12)
plt.plot(x1,y1,label="line-2",alpha=0.1)
plt.legend()
plt.show()
```

```python
plt.bar(df['class'],df['sepal_length'],color="purple",width=0.7,edgecolor="red",label="iris")
plt.legend()
plt.title("Students Marks in Semester")
plt.show()
```

```python
plt.scatter(df['petal_length'],df['petal_width'],color="purple",marker="*",s=200,label="iris")
plt.legend()
plt.show()
```
