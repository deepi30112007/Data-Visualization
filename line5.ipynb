```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[2,4,6,9,15]
x1=[6,7,8,9]
y1=[9,5,10,8]
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("line plot Example")
plt.plot(x,y,label="line-1",color='purple',linestyle='dashdot',linewidth=3,marker='D',markerfacecolor='red',markersize=12)
plt.plot(x1,y1,label="line-2",alpha=0.1)
plt.legend()
plt.savefig("sample.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
students=["Deepika","Anu","Nethra","Saranya","Nafisa"]
marks=[60,50,40,30,65]
plt.bar(students,marks,color="purple",width=0.7,edgecolor="red",label="StudentsMarks")
plt.legend()
plt.title("Students Marks in Semester")
plt.xlabel("students")
plt.ylabel("marks")
plt.show()
```

```python
Hours=[2,3,4,5,6]
Marks=[45,55,65,70,80]
plt.scatter(Hours,Marks,color="blue",marker="D",s=200,label="StudentsMarks")
plt.legend()
plt.show()
```

```python
marks=[50,60,60,70,75,70,80,85,90]
plt.hist(marks,bins=8,color='purple',edgecolor="blue",label="Marks")
plt.title("Histogram")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend()
plt.show()
```

```python
import pandas as pd
mk=pd.Series(marks)
mk
```

```python
mk.plot(kind="density",color="purple",linewidth=2,linestyle="--",label="Marks")
plt.legend()
plt.grid(True)
plt.show()
```
