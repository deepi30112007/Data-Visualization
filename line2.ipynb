```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[2,4,6,9,15]
x1=[6,7,8,9]
y1=[9,5,10,8]
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("line plot Example")
plt.plot(x,y,label="line-1",color='purple',linestyle='dashdot',linewidth=3,marker='D',markerfacecolor='red',markersize=12)
plt.plot(x1,y1,label="line-2",color='orange',linestyle='dashdot',linewidth=3,marker='o',markerfacecolor='yellow',markersize=12)
plt.legend()
plt.savefig("sample.png")
plt.show()
```
