```python
#Traversal
arr=[40,50,60,70,80]
for element in arr:
    print(element)
```

```python
#Insertion
arr=[7,6,4,8]
arr.insert(2,40)
print(arr)
```

```python
#Deletion
arr=[1,2,3,4,5]
arr.remove(3)
print(arr)
```

```python
#Searching
arr=[10,20,30,40,50]
key=40
if key in arr:
    print("Element found")
else:
    print(" Element Not found")
```

```python
#Update
arr=[6,7,8,9,10]
arr[2]=40
print(arr)
```

```python
#Sorting
arr=[30,40,70,60,50]
arr.sort()
print(arr)
```

```python
# Searching
arr = [10, 20, 30, 40, 50]
key = 40

for i in arr:
    if i == key:
        print("Element found")
        break
else:
    print("Element not found")
```
